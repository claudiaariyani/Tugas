```python
class Dosen:
    def __init__(self, nama, nidn, gaji):
        self.nama = nama            # Atribut Public
        self._nidn = nidn          # Atribut Protected
        self.__gaji = gaji          # Atribut Private yang dilindungi

    # Getter menggunakan @property untuk mengakses atribut private __gaji
    @property
    def gaji(self):
        return self.__gaji

    # Setter menggunakan @gaji.setter untuk memvalidasi dan mengubah __gaji
    @gaji.setter
    def gaji(self, gaji_baru):
        if gaji_baru > 0:
            self.__gaji = gaji_baru


    def tampilkan_data(self):
        # Memanggil atribut gaji langsung melalui property getter
        print(f"Nama : {self.nama}")
        print(f"NIDN : {self._nidn}")
        print(f"Gaji : Rp{self.gaji:,}")
        print("-" * 30)


# Membuat 2 objek
dosen1 = Dosen("Dr. Budi Santoso, S.Kom., M.Kom.", "0512038001", 8500000)
dosen2 = Dosen("Rena Maryani, S.Kom., M.Cs.", "0515088902", 9000000)

print("=== INFORMASI DOSEN 1 ===")
dosen1.tampilkan_data()

print("=== INFORMASI DOSEN 2 ===")
dosen2.tampilkan_data()

print("=== PENGGUNAAN GETTER DAN SETTER ===")
print(f"Gaji lama Dosen 1: Rp{dosen1.gaji:,}")  # Mengakses getter

# Mengakses setter
dosen1.gaji = 9500000
print(f"Gaji baru Dosen 1: Rp{dosen1.gaji:,}")
```

```python
class Mahasiswa:
    def __init__(self, nama, npm):
        self.nama = nama        # atribut public
        self.npm = npm          # atribut public
        self.__nilai = []       # atribut private berupa list

    # Getter
    @property
    def nilai(self):
        return self.__nilai

    # Setter untuk menambahkan nilai baru ke dalam list
    @nilai.setter
    def nilai(self, nilai_baru):
        if 0 <= nilai_baru <= 100:   # validasi rentang 0-100
            self.__nilai.append(nilai_baru)
        else:
            raise ValueError(f"Nilai {nilai_baru} tidak valid. Harus di rentang 0-100.")

    # Metode menghitung rata-rata nilai
    def rata_nilai(self):
        if not self.__nilai:
            return 0.0
        return sum(self.__nilai) / len(self.__nilai)

    def tampilkan_info(self):
        print(f"Nama         : {self.nama}")
        print(f"NPM          : {self.npm}")
        print(f"Daftar Nilai : {self.nilai}")
        print(f"Rata-rata    : {self.rata_nilai():.2f}")

    # menghitung grade
    def hitung_grade(self):
        rata = self.rata_nilai()
        if rata >= 85:
            return "A"
        elif rata >= 75:
            return "B"
        elif rata >= 60:
            return "C"
        elif rata >= 50:
            return "D"
        else:
            return "E"

    def status_kelulusan(self):
        if self.hitung_grade() in ("A", "B", "C"):
            return "Lulus"
        return "Tidak Lulus"

    def tampilkan_transkrip(self):
        print("=" * 40)
        print("          TRANSKRIP NILAI")
        print("=" * 40)
        print(f"Nama             : {self.nama}")
        print(f"NPM              : {self.npm}")
        print(f"Daftar Nilai     : {self.nilai}")
        print(f"Rata-rata        : {self.rata_nilai():.2f}")
        print(f"Grade            : {self.hitung_grade()}")
        print(f"Status Kelulusan : {self.status_kelulusan()}")
        print("=" * 40)
        print()


#objek mahasiswa

# Mahasiswa 1 -> Lulus
mhs1 = Mahasiswa("Siti Rahmawati", "5230411002")

#menambahkan nilai menggunakan setter
mhs1.nilai = 90
mhs1.nilai = 90
mhs1.nilai = 78

# Mahasiswa 2 -> Tidak Lulus
mhs2 = Mahasiswa("Seonghyeon Cortis", "5230411003")

# menambahkan nilai menggunakan setter
mhs2.nilai = 55
mhs2.nilai = 48
mhs2.nilai = 52

# Uji validasi menggunakan setter
print("=== UJI VALIDASI NILAI ===")
for nilai_uji in (105, -5):
    try:
        mhs1.nilai = nilai_uji
    except ValueError as e:
        print(f"Input {nilai_uji} ditolak -> {e}")
print(f"Nilai mhs1 tetap aman: {mhs1.nilai}\n")

# tampilkan transkrip
mhs1.tampilkan_transkrip()
mhs2.tampilkan_transkrip()
```
