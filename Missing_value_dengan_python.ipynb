8. Medeteksi angka(sebagai missing value) pada kolom yang bertipe non numerik dan cekdata pada kolom St_Milik(no 3). Apa artinya?


Artinya....

```python
# Deteksi angka
cnt = 0

for row in df['St_Milik']:
    try:
        int(row)
        df.loc[cnt, 'St_Milik'] = np.nan
    except (ValueError, TypeError):
        pass

    cnt += 1

df['St_Milik'].isnull()
```

10. Mengisi missing value dengan angka pada lokasi yang spesifik

```python
# Mengganti missing values dengan sebuah angka
df['KodeApt'].fillna(837, inplace=True)
print(df.KodeApt)
```

```python
#Mengganti pada lokasi spesifik
df.loc[2, 'KodeApt'] = 8837
print(df.KodeApt)
```

11. Mengisi missing value dengan nilai median


```python
#Mengganti missing value dengan median
median = df['Jum_kamar']. median()
df['Jum_kamar'].fillna(median, inplace=True)
print(df.Jum_kamar)
```

15. Mengubah data non numerik menjadi numerik


```python
# Libraries
from sklearn.preprocessing import LabelEncoder
from pandas import DataFrame
import pandas as pd
import numpy as np

# Melihat semua isi data
print(df)
le = LabelEncoder()

for col in df.columns.values:
    # Encoding pada variable kategoriyal16
    if df[col].dtypes == 'object':
        # Menggunakan pd.concat sebagai pengganti .append yang sudah deprecated/dihapus
        data = pd.concat([df[col], df[col]])
        le.fit(data.values)
        df[col] = le.transform(df[col])

df.head(10)
```
